## Swapping MultiIndex Levels

```python
data = {
    "Dept":["CSE", "ECE", "CSE","ECE","CSE","ECE"],
    "Student_ID":[103,101,102,203,201,202],
        "Name":["Aman","Riya","Ujjwal","Yuvraj","Yash","Shubham"],
    "Marks":[90,80,70,60,50,40]
}
df=pd.DataFrame(data)
df

```

```python
muti_df1=df.sort_index()
```

```python
muti_df1.swaplevel()
```

```python
muti_df1 = df.set_index(["Dept","Student_ID"])
Swapped_df = muti_df1.swaplevel("Dept","Student_ID")
print(Swapped_df)
```

```python
print(Swapped_df.sort_index())
```

## Combining and Merging Datasets

```python
class_a = pd.DataFrame({
    "Student_ID":[1,2],
    "Name":["Aman","Ujjwal"],
    "Marks":[90,80]
})
class_b = pd.DataFrame({
    "Student_ID":[3,4],
    "Name":["Yash","Yuvraj"],
    "Marks":[70,60]
})
all_students = pd.concat([class_a,class_b],ignore_index=True)
print(all_students)
```

```python
all_students = pd.concat([class_a,class_b],axis=0)
print(all_students)
```

```python
all_students = pd.concat([class_a,class_b],axis=1)
print(all_students)
```

## Merge

```python
student = pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Aman","Ujjwal","Yash","Yuvraj"],
})
result = pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Marks":[92,82,75,61]
})
```

```python
merges = pd.merge(student,result,on="Student_ID",how="left")
print(merges)
```

```python
merges = pd.merge(student,result,on="Student_ID",how="outer")
print(merges)
```

```python
merges = pd.merge(student,result,on="Student_ID",how="right")
print(merges)
```

```python
merges = pd.merge(student,result,how="cross")
print(merges)
```

## Merging When key-column names differ

```python
student_info = pd.DataFrame({
    "Student_ID":[1,2,3,4],
    "Name":["Aman","Ujjwal","Yash","Yuvraj"],
})
result_info = pd.DataFrame({
    "ID":[1,2,3,5],
    "Marks":["92","82","75","61"]
})
answer = pd.merge(student_info,result_info,left_on="Student_ID",right_on="ID")
print(answer)
```

## Reshaping and Pivoting

```python
wide = pd.DataFrame({
    "Name":["Aman","Ujjwal","Yash","Yuvraj"],
    "Maths":[85,90,78,93],
    "Science":[84,88,93,89],
    "English":[82.89,64,89,85]
})
print(wide)
```

## Convert wide data to long using melt()

```python
wide = pd.DataFrame({
    "Name":["Riya","Aman","Yuvraj","Ujjwal"],
    "Maths":[85,90,78,93],
    "Science":[84,88,93,89],
    "English":[82,89,84,86]
})
print(wide)
```

## Convert long data to wide using pivot()

```python
# First, create the 'long' DataFrame from 'wide' using melt()
long = pd.melt(
    wide,
    id_vars=["Name"],
    value_vars=["Maths", "Science", "English"],
    var_name="Subject",
    value_name="Marks"
)

# Now, pivot the 'long' DataFrame back to wide format
wide_again = long.pivot(
    index="Name",
    columns="Subject",
    values="Marks"
)
print(wide_again)
```

```python
sales = pd.DataFrame({
    "City":["Agra","Agra","Delhi","Delhi","Agra"],
    "Product":["Pen","Pen","Pen","Pencil","Pencil"],
    "Sales":[100,200,150,300,250]
})
summary = sales.pivot_table(
    index = "City",
    columns = "Product",
    values = "Sales",
    aggfunc = "sum",
    fill_value=0
)
print(summary)
```

## Crosstabulation

```python
data = pd.DataFrame({
    "Dept":["CSE","CSE","ECE","ECE","CSE"],
    "Result":["Pass","Fail","Pass","Pass","Pass"],
})
table = pd.crosstab(data["Dept"],data["Result"])
print(table)
```

## Adding totals and percentage

```python
print(pd.crosstab(data["Dept"],data["Result"],margins=True))
```

```python
print(pd.crosstab(data["Dept"],data["Result"],normalize="index"))
```
