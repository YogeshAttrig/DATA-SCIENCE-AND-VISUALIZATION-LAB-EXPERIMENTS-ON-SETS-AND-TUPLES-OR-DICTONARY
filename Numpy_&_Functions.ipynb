## Reshaping Array

```python
data = np.arange(1,13)
a= data.reshape(3,4)
b= data.reshape(4,3)
c= data.reshape(2,-1)
print(f'1D:{data}')
print(f'3x4:\n{a}')
print(f'4x3:\n{b}')
print(f'2xinfer:\n{c}')

print(f'A shape:\n{a.shape}')
print(f'B shape:\n{b.shape}')
print(f'C shape:\n{c.shape}')
```

Make a function to take input from user and reshape numpy array into 6x2 matrix 3x-1 matrix and 5x-1 matrix. If error comes please explain the error

```python
def reshape_user_input():
    try:
        arr = np.array([float(x) for x in input("Numbers: ").split()])
    except ValueError:
        print("Error: Invalid numbers input.")
        return

    n = arr.size
    for shape in [(6, 2), (3, -1), (5, -1)]:
        try:
            print(f"Shape {shape}:\n{arr.reshape(shape)}\n")
        except ValueError:
            req = shape[0] if shape[1] == -1 else shape[0] * shape[1]
            if shape[1] == -1:
                print(f"Error for {shape}: Total elements ({n}) must be divisible by {req}.\n")
            else:
                print(f"Error for {shape}: Requires exactly {req} elements, got {n}.\n")

reshape_user_input()
```
