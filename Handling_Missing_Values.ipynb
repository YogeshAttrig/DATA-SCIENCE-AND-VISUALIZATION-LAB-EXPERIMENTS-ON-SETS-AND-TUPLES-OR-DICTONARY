## Simple Statistical Imputation

Mean Imputation

```python
from numpy._core.fromnumeric import mean
from sklearn.impute import SimpleImputer

mean_imputer = SimpleImputer(strategy="mean")
clean = students.copy()
clean["Marks"] = clean["Marks"].fillna(clean["Marks"].mean())
clean["Attandance"] = clean["Attendance"].fillna(clean["Attendance"].mean())
clean["Name"] = clean["Name"].fillna("Unknown")
print(clean[["Name","Marks","Attendance"]])
```

```python
from numpy._core.fromnumeric import mean
from sklearn.impute import SimpleImputer
df_mean = students.copy()
mean_imputer = SimpleImputer(strategy="mean")
clean = students.copy()
df_mean[["Marks","Attendance"]] = mean_imputer.fit_transform(df_mean[["Marks","Attendance"]])
df_mean["Name"] = df_mean["Name"].fillna("Unknown")
print(df_mean[["Name","Marks","Attendance"]])
```

```python

from sklearn.impute import SimpleImputer
df_median = students.copy()
median_imputer = SimpleImputer(strategy="median")
clean = students.copy()
df_median[["Marks","Attendance"]] = median_imputer.fit_transform(df_median[["Marks","Attendance"]])
df_median["Name"] = df_median["Name"].fillna("Unknown")
print(df_median[["Name","Marks","Attendance"]])
```

```python

from sklearn.impute import SimpleImputer
df_mode = students.copy()
mode_imputer = SimpleImputer(missing_values=None, strategy="most_frequent")
clean = students.copy()
df_mode[["Name"]] = mode_imputer.fit_transform(df_mode[["Name"]])
print(df_mode)
```

```python
from sklearn.impute import SimpleImputer
df_const = students.copy()
constant_imputer = SimpleImputer(strategy="constant",fill_value=0)
df_const[["Marks","Attendance"]] = constant_imputer.fit_transform(df_mode[["Marks","Attendance"]])
print(df_const)
```

## KNN Imputation

```python
from sklearn.impute import KNNImputer
df_knn = students.copy()
knn_imputer = KNNImputer(n_neighbors=2)
df_knn[["Marks","Attendance"]] = knn_imputer.fit_transform(df_knn[["Marks","Attendance"]])
df_knn["Name"] = df_knn["Name"].fillna("Unknown")
print(df_knn)
```

## Remove Duplicates using an ID

```python
unique_student = no_exact_duplicates.drop_duplicates(
    subset=["Student_ID"], keep = "first"
)
print("Rows before:", len(clean))
print("Rows after:", len(unique_student))
print("After Removal:", unique_student.shape[0])
unique_student
```

## Fix inconsistent text

```python
unique_student = unique_student.copy()
unique_student["Name"] = unique_student["Name"].str.strip().str.title()
print(unique_student["Name"])
```

```python
unique_student["Dept"] = unique_student["Dept"].str.strip().str.lower()
print(unique_student["Dept"])
```

```python
dept_map ={
     "cse":"CSE",
     "ece":"ECE",
     "dataa science":"Data Science",

}
unique_student["Dept"]=unique_student["Dept"].map(dept_map).fillna("Needs review")
print(unique_student["Dept"])
```

## Convert dates and Check ranges

```python
unique_student["Exam_Date"] = pd.to_datetime(unique_student["Exam_Date"], errors ="coerce", dayfirst=True)
print(unique_student["Exam_Date"])
```

```python
print("Missing or invalid dates:",unique_student["Exam_Date"].isnull().sum())
```

```python
bad_marks = unique_student[~unique_student["Marks"].between(0,100)]
print("Marks outside 0-100")
print(bad_marks)
```
